# Evaluating a prediction interval

In practice, producing a prediction interval that offers the correct level of coverage is tricky. If the model is a poor fit then it will produce PIs that are too wide. While more complex methods tend to be overconfident and produce intervals that are too narrow.  `forecast-tools.metrics` provides three functions to support evaluating prediction intervals.

* `coverage` - a basic measure of the proportion of observations covered by the interval
* `absolute_coverage_difference` - the deviation of empirical coverage from that expected.
* `winker_score` - a scale dependent measure equal to the width of the interval plus a penality proportional to the deviation (above or below the interval) and 2/$\alpha$

## 1. Imports

```python
import pandas as pd
import numpy as np

from forecast_tools.baseline import SNaive, Naive1
from forecast_tools.metrics import (
    winkler_score,
    absolute_coverage_difference,
    coverage,
    interval_scores
)
from forecast_tools.datasets import load_emergency_dept
```

## 2. Empirical Coverage

A simple way evaluate a prediction interval is to quantify its empirical coverage of a test set.

The intervals in the example below have 83% coverage of the true values.

```python
intervals = np.array(
    [
        [37520, 58225],
        [29059, 49764],
        [47325, 68030],
        [36432, 57137],
        [35865, 56570],
        [33419, 54124],
    ]
)

y_true = np.array([37463, 40828, 56148, 45342, 43741, 45907])
mean_cov = coverage(y_true, intervals)
print(round(mean_cov, 2))
```

## 3. Absolute Coverage Difference (ACD)

You can enhance the basic coverage metric by considering a desired (or target) coverage that the method promised.  The absolute coverage difference is the absolute difference between average coverage and the target coverage level.

For example if the average coverage is 83% and the target is a 95% coverage then the absolute coverage difference is

$|0.83 - 0.95| = 0.12$

```python
intervals = np.array(
    [
        [37520, 58225],
        [29059, 49764],
        [47325, 68030],
        [36432, 57137],
        [35865, 56570],
        [33419, 54124],
    ]
)

y_true = np.array([37463, 40828, 56148, 45342, 43741, 45907])
acd = absolute_coverage_difference(y_true, intervals, alpha=0.05)
print(round(acd, 2))
```

## 4. Winkler score (also called interval score)

An alternative way to assess prediction interval coverage for a specific time series is using the Winkler score. 

The winkler score is defined as

$$W_{\alpha,t} = \begin{cases}
  (u_{\alpha,t} - \ell_{\alpha,t}) + \frac{2}{\alpha} (\ell_{\alpha,t} - y_t) & \text{if } y_t < \ell_{\alpha,t} \\
  (u_{\alpha,t} - \ell_{\alpha,t})   & \text{if }  \ell_{\alpha,t} \le y_t \le u_{\alpha,t} \\
  (u_{\alpha,t} - \ell_{\alpha,t}) + \frac{2}{\alpha} (y_t - u_{\alpha,t}) & \text{if } y_t > u_{\alpha,t}.
  \end{cases}$$
  
  
Where 

* $u_{\alpha, t}$ is the upper prediction interval value for $\alpha$ and horizon $t$
* $l_{\alpha, t}$ is the lower prediction interval value for $\alpha$ and horizon $t$
* $y_t$ is the ground truth observation at horizon $t$

**sources**

* https://otexts.com/fpp3/distaccuracy.html
* https://www.tandfonline.com/doi/abs/10.1198/016214506000001437
* https://www.tandfonline.com/doi/pdf/10.1080/01621459.1972.10481224?needAccess=true

### Interpretation

* A Winkler score is the width of the interval plus a penality proportional to the deviation (above or below the interval) and 2/$\alpha$
* Smaller winkler scores are better!

### 4.1 Winkler score for a 1-step forecast

Assume our model has generated a prediction interval of [744.54, 773.22].  We know that the true observation lies outside of the lower interval 741.84.  The winkler score is therefore:

```python
alpha = 0.2
interval = [744.54, 773.22]
y_t = 741.84
ws = winkler_score(y_t, interval, alpha)
print(round(ws, 2))
```

if instead our $y_t$ fell within the interval.  The winkler score would be equal to the width of the interval.

```python
interval = [744.54, 773.22]
y_t = 745.0
ws = winkler_score(y_t, interval, alpha)
print(f"Winkler score: {round(ws, 2)}")
print(f"Interval width: {round(interval[1]-interval[0], 2)}")
```

### 4.2 Winkler score for multiple step forecast

More commonly a model will generate a prediction over a multi-step horizon.  Here for example Seasonal naive is used to predict a 7 day holdout sample in the ED dataset.

```python
HOLDOUT = 7
PERIOD = 7

attends = load_emergency_dept()

# train-test split
train, test = attends[:-HOLDOUT], attends[-HOLDOUT:]

model = SNaive(PERIOD)

# returns 80 and 90% prediction intervals by default.
preds, intervals = model.fit_predict(train, HOLDOUT, return_predict_int=True)

ws = winkler_score(test, intervals[0], alpha=0.2)
print(f"Winkler score: {ws:.2f}")
```

We can view the individual winkler scores within the interval by setting the `return_scores` parameter to true. The function will return a `dict` instead of a `float`

```python
ws = winkler_score(test, intervals[0], alpha=0.2, return_scores=True)
# the individual winkler scores are store with key "individual scores"
print(ws['individual_scores'])
```

## 5. Formatting intervals to pass to the evaluation functions

The functions have been designed to be as accomodating as possible to your data types. For an m-step forecast the expectation is that the software you have used has generated a `pandas.DataFrame` or a `numpy.ndarray`

### 5.1. array format

The code below takes a closer look at the format of the 80% intervals returned from `SNaive`. This is a `numpy.ndarray` with shape = (n, 2) where n is the forecast horizon.

```python
HOLDOUT = 7
PERIOD = 7

attends = load_emergency_dept()

# train-test split
train, test = attends[:-HOLDOUT], attends[-HOLDOUT:]

model = SNaive(PERIOD)

# returns 80 and 90% prediction intervals by default.
preds, intervals = model.fit_predict(train, HOLDOUT, return_predict_int=True)

intervals_80 = intervals[0]
intervals_90 = intervals[1]

print(type(intervals_80))
print(intervals_80.shape)
print(intervals_80)
```

```python
ws = winkler_score(test, intervals_80, alpha=0.2)
print(f"Winkler score: {ws:.2f}")
```

```python
cover = coverage(test, intervals_80)
print(f"Coverage: {cover:.2f}")
```

```python
acd = absolute_coverage_difference(test, intervals_80, alpha=0.2)
print(f"Absolute Coverage Difference: {acd:.2f}")
```

### 5.2 Dataframe format

Alternatively your intervals might be stored in a `pandas.DataFrame`. The functions can handle this format by default  The code below converts the 80% prediction intervals to a dataframe.  Take a close look at the the output.  Note that the shape is again (n, 2) where n is the forecast horizon.

```python
# convert  80% interfvals to pandas dataframe
intervals_80 = pd.DataFrame(
    intervals[0], 
    index=test.index, 
    columns=["upper", "lower"]
)

print(intervals_80.shape)
intervals_80.round(1)
```

We can simply pass the `DataFrame` to the interval evaluation functions.

```python
ws = winkler_score(test, intervals_80, alpha=0.2)
print(f"Winkler score: {ws:.2f}")
```

```python
cover = coverage(test, intervals_80)
print(f"Coverage: {cover:.2f}")
```

```python
acd = absolute_coverage_difference(test, intervals_80, alpha=0.2)
print(f"Absolute Coverage Difference: {acd:.2f}")
```

## 6. Interval score kitchen sink function

`forecast-tools.metrics` includes a function called `interval_scores` that will return all available interval scores to the user for a given set of intervals, and observations.  The function returns a `dict`.

```python
HOLDOUT = 7
PERIOD = 7

attends = load_emergency_dept()

# train-test split
train, test = attends[:-HOLDOUT], attends[-HOLDOUT:]

model = SNaive(PERIOD)

# returns 80 and 90% prediction intervals by default.
preds, intervals = model.fit_predict(train, HOLDOUT, return_predict_int=True)

intervals_80 = intervals[0]

scores = interval_scores(test, intervals_80, alpha=0.2)
scores
```

```python
# limit which interval scores are returned
scores = interval_scores(
    test, intervals_80, alpha=0.2, metrics=["acd", "winkler"])
scores
```
